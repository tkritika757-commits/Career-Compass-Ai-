```python
import pandas as pd
from itertools import product

subjects = {
    "Biology": ["Helping people", "Research"],
    "Physics": ["Problem solving", "Research", "Technology"],
    "Chemistry": ["Research", "Helping people"],
    "Mathematics": ["Problem solving", "Technology"],
    "Computer Science": ["Technology", "Problem solving"],
    "Business Studies": ["Leadership", "Communication"],
    "English": ["Communication", "Creativity"],
    "Fine Arts": ["Creativity", "Communication"]
}

work_styles = {
    "Helping people": ["People", "Research"],
    "Research": ["Research", "Technology"],
    "Technology": ["Technology", "Research"],
    "Problem solving": ["Technology", "Research"],
    "Leadership": ["Business", "People"],
    "Communication": ["People", "Creative"],
    "Creativity": ["Creative", "People"]
}

career_map = {
    ("Biology", "Helping people", "People"): "Healthcare",
    ("Biology", "Research", "Research"): "Research",
    ("Chemistry", "Research", "Research"): "Research",
    ("Physics", "Problem solving", "Technology"): "Engineering",
    ("Mathematics", "Problem solving", "Technology"): "Data and Analytics",
    ("Computer Science", "Technology", "Technology"): "Technology",
    ("Business Studies", "Leadership", "Business"): "Business",
    ("English", "Communication", "People"): "Communication",
    ("Fine Arts", "Creativity", "Creative"): "Creative Industries"
}

rows = []

for subject, interests in subjects.items():
    for interest in interests:
        for style in work_styles[interest]:
            career = career_map.get(
                (subject, interest, style),
                {
                    "Biology": "Healthcare",
                    "Physics": "Engineering",
                    "Chemistry": "Research",
                    "Mathematics": "Data and Analytics",
                    "Computer Science": "Technology",
                    "Business Studies": "Business",
                    "English": "Communication",
                    "Fine Arts": "Creative Industries"
                }[subject]
            )

            rows.append({
                "subject": subject,
                "interest": interest,
                "work_style": style,
                "career": career
            })

df = pd.DataFrame(rows).drop_duplicates().reset_index(drop=True)

print("CareerCompass AI dataset created!")
print("Number of examples:", len(df))
display(df.head(10))
print("\nExamples in each career category:")
display(df["career"].value_counts())
```

```python
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, classification_report

features = ["subject", "interest", "work_style"]
X = df[features]
y = df["career"]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=42
)

preprocessor = ColumnTransformer([
    ("categorical", OneHotEncoder(handle_unknown="ignore"), features)
])

model = Pipeline([
    ("preprocessor", preprocessor),
    ("classifier", DecisionTreeClassifier(
        max_depth=4,
        random_state=42
    ))
])

model.fit(X_train, y_train)

predictions = model.predict(X_test)

print("CareerCompass AI retrained successfully!")
print("Training examples:", len(X_train))
print("Testing examples:", len(X_test))
print("Test accuracy:", round(accuracy_score(y_test, predictions), 2))
print("\nClassification report:")
print(classification_report(y_test, predictions, zero_division=0))
```
